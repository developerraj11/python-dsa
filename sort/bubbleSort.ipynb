```python
def bubble_sort(arr):
	n = len(arr)

	for end in range(n - 1, 0, -1):
		swapped = False

		for index in range(end):
			if arr[index] > arr[index + 1]:
				arr[index], arr[index + 1] = arr[index + 1], arr[index]
				swapped = True

		if not swapped:
			break

	return arr
```

```python
print(bubble_sort([3,1,4,5,7,2,9,8]))
```
